
## Punto 1 — 3 KPIs que MercadoLibre debería seguir para medir el crecimiento

### KPI 1 — % de personas con internet por país
*El más importante*

En 2020 el promedio de penetración de internet en LATAM es del 68.9%
sobre una población total de 599 millones de personas. Eso significa
que aproximadamente 186 millones de personas en LATAM todavía no
tienen acceso a internet, por tanto no pueden ser clientes de
MercadoLibre.

Este es el KPI #1 porque es el habilitador directo del e-commerce.
Sin internet no hay usuario, sin usuario no hay compra, sin compra
no hay negocio. Cada punto porcentual que sube la penetración de
internet en LATAM representa aproximadamente 6 millones de nuevos
usuarios digitales potenciales para MercadoLibre.

Los datos muestran además que la penetración no es uniforme.
Chile lidera con 88.3% mientras Honduras tiene 42%. Esa brecha
define dónde MercadoLibre tiene mercado maduro y dónde tiene
oportunidad de expansión.



### KPI 2 — Suscripciones a telefonía celular móvil por país

En 2020 LATAM tiene 775 millones de suscripciones celulares sobre
una población de 599 millones, una razón de 1.29 suscripciones
por persona. Eso significa que el celular ya llegó a prácticamente
toda la población, incluso a segmentos que aún no tienen internet.

Este es el KPI #2 porque anticipa el crecimiento del KPI #1.
Históricamente en LATAM la secuencia es primero celular, luego
internet móvil. Un país con alta penetración celular y baja
penetración de internet está en la antesala de la digitalización
masiva — es el momento exacto en que MercadoLibre debería
invertir en adquisición de usuarios antes de que llegue la
competencia.




### KPI 3 — Población total por país

LATAM tiene 599 millones de personas. Brasil concentra 213 millones
y México 126 millones, entre los dos suman el 56% de la población
regional. Colombia tiene 51 millones y Perú 33 millones.

Este es el KPI #3 porque contextualiza los otros dos. Un 70% de
penetración de internet en Brasil representa 149 millones de
usuarios digitales. El mismo 70% en Uruguay representa apenas
2.4 millones. Sin la población como denominador, los porcentajes
no tienen dimensión de negocio.

La población también define la prioridad de inversión. Colombia
con 51 millones de habitantes y 69.8% de internet tiene más
usuarios digitales absolutos que Chile con 88.3% de internet
pero solo 19 millones de habitantes.
